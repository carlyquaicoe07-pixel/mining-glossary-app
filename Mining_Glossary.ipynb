```python
# STEP 3 — SEARCH THE GLOSSARY

while True:
    print("")
    term = input("Enter a mining term (or 'quit' to exit): ")

    if term.lower() == "quit":
        print("Thank you for using the Mining Glossary App!")
        break

    if term.lower() in glossary:
        print(f"Definition: {glossary[term.lower()]}")
    else:
        print(f"Term '{term}' not found. Try another term.")

# SAVE TO FILE
with open("mining_glossary.txt", "w") as f:
    f.write("MINING GLOSSARY\n")
    f.write("Built by Carly Edward Quaicoe — UMaT\n")
    f.write("="*40 + "\n")
    for term, definition in glossary.items():
        f.write(f"{term.upper()}: {definition}\n")

print("Glossary saved to mining_glossary.txt")
```
